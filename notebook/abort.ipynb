```python
import numpy as np
import multiprocess as mp
import matplotlib.pyplot as plt
import matplotlib as mpl

from discrete_simulation import DiscreteSimulation
```

```python
d = 16
m_i = 16
m_c = 16
fidelity_cut = 0
gamma_list = np.logspace(-3,-1,41)
gamma_ratio = 1
alpha=7

process_count = mp.cpu_count()-1

simulation16 = DiscreteSimulation(d, m_i,m_c, alpha=alpha )

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation16.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list1616 = [tup[0] for tup in fid_prob_list]
prob_list1616 = [tup[1] for tup in fid_prob_list]

print(1)

m_i = 16
m_c = 8
simulation8 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation8.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_168 = [tup[0] for tup in fid_prob_list]
prob_list_168 = [tup[1] for tup in fid_prob_list]

print(2)

m_i = 16
m_c = 4
simulation4 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation4.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_164 = [tup[0] for tup in fid_prob_list]
prob_list_164 = [tup[1] for tup in fid_prob_list]

print(3)

m_i = 16
m_c = 2
simulation2 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation2.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_162 = [tup[0] for tup in fid_prob_list]
prob_list_162 = [tup[1] for tup in fid_prob_list]


print(4)
```

```python
d = 32
m_i = 16
m_c = 16
fidelity_cut = 0.9
gamma_list = np.logspace(-3,-1,41)
gamma_ratio = 1
alpha=7

process_count = mp.cpu_count() - 1

simulation16 = DiscreteSimulation(d, m_i,m_c, alpha=alpha )

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation16.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list1616cut = [tup[0] for tup in fid_prob_list]
prob_list1616cut = [tup[1] for tup in fid_prob_list]

print(1)

m_i = 16
m_c = 8
simulation8 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation8.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_168cut = [tup[0] for tup in fid_prob_list]
prob_list_168cut = [tup[1] for tup in fid_prob_list]

print(2)

m_i = 16
m_c = 4
simulation4 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation4.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_164cut = [tup[0] for tup in fid_prob_list]
prob_list_164cut = [tup[1] for tup in fid_prob_list]
```

```python
m_i = 16
m_c = 2
simulation2 = DiscreteSimulation(d, m_i,m_c, alpha=alpha)

with mp.Pool(process_count) as pool:
    fid_prob_list = pool.starmap(simulation2.average_fidelity, zip(list(gamma_list),list(gamma_ratio*gamma_list), [fidelity_cut] * len(gamma_list)))
fid_list_162cut = [tup[0] for tup in fid_prob_list]
prob_list_162cut = [tup[1] for tup in fid_prob_list]
```

```python
plt.style.use('bmh')
mpl.rcParams['font.family'] = 'DejaVu Serif'
mpl.rcParams["axes.facecolor"] = "w"
# ax = plt.axes()
# ax.set_facecolor("w")

mpl.rc('xtick', labelsize=14) 
mpl.rc('ytick', labelsize=14) 

plt.figure(dpi=100, figsize = (6.4,3.5))
plt.rcParams['font.size'] = 16

best_mc = np.minimum.reduce([1-np.array(fid_list1616), 
                     1-np.array(fid_list_168),
                     1-np.array(fid_list_164),
                     1-np.array(fid_list_162)
                     ])

fid_list_list = [1-np.array(fid_list1616cut), 
                     1-np.array(fid_list_168cut),
                     1-np.array(fid_list_164cut),
                     1-np.array(fid_list_162cut)
                     ]
prob_list_list = [np.array(prob_list1616cut), 
                     np.array(prob_list_168cut),
                     np.array(prob_list_164cut),
                     np.array(prob_list_162cut)
                     ]


best_cut = []
cut_prob = []
for i in range(len(prob_list1616)):
    fids = [fid_list[i] for fid_list in fid_list_list]
    probs = [prob_list[i] for prob_list in prob_list_list]
    # print(probs)
    best_cut.append(np.min(fids))
    # print(np.argmin(fids))
    cut_prob.append(probs[np.argmin(fids)])

nonzero_indices = np.where(np.array(best_cut) != 1)[0]

plt.loglog(gamma_list,best_mc, '-', label="$f_{\\mathrm{cut}}=0$")
plt.loglog(gamma_list[nonzero_indices],np.array(best_cut)[nonzero_indices], '-', label="$f_{\\mathrm{cut}}=0.9$")
plt.loglog(gamma_list[nonzero_indices],np.array(cut_prob)[nonzero_indices], '--', color= u'#A60628', label="$p_\\mathrm{abort}$ for $f_{\\mathrm{cut}}=0.9$")


# plt.axhline(0.5, color = 'r', linestyle = ':')


# plt.title("Infidelity with trivial protocol and with our protocol. d = 16, m_i = 4, m_c=4")
plt.xlabel(f"$\\gamma$", fontsize=14)  # decay and 1/{gamma_ratio} 
plt.ylabel("Infidelity", fontsize=14)
plt.ylim([1,1e-6])
plt.xlim([1e-3,1e-1])
plt.legend(fontsize=14)
# plt.title(f"${d=},{m_i=}, \\alpha={alpha} $")
# plt.gca().invert_yaxis()

# Create a new y-axis that shares the x-axis
# ax2 = plt.gca().twinx()

# Plot `cut_prob` on the new y-axis
# ax2.loglog(gamma_list[nonzero_indices], np.array(cut_prob)[nonzero_indices], '--', color='#A60628', label="$p_\\mathrm{fail}$ for $f_{\\mathrm{cut}}=0.9$")

# # Set the new y-axis label and adjust its limits if necessary
# ax2.set_ylabel("Probability", fontsize=14)
# ax2.set_ylim([1, 1e-3])

plt.axvspan(gamma_list[nonzero_indices.max()], 1, color='red', alpha=0.1)

plt.text(0.05, 0.7*1e-2, "$p_\\mathrm{abort}=1$", fontsize=14, ha='center', va='center', color='black')

plt.savefig(f'../plots/f_cut,{d=},{alpha=}.svg', bbox_inches='tight')
plt.show()
```

```python
mc_series = [
    (16, 1 - np.array(fid_list1616cut), np.array(prob_list1616cut)),
    (8, 1 - np.array(fid_list_168cut), np.array(prob_list_168cut)),
    (4, 1 - np.array(fid_list_164cut), np.array(prob_list_164cut)),
    (2, 1 - np.array(fid_list_162cut), np.array(prob_list_162cut)),
]

plt.style.use('bmh')
mpl.rcParams['font.family'] = 'DejaVu Serif'
mpl.rcParams['axes.facecolor'] = 'w'
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)
plt.rcParams['font.size'] = 13

fig, axes = plt.subplots(2, 2, figsize=(8.0, 5.4), dpi=120, sharex=True)
axes = axes.ravel()


for ax, (mc, infid, abort_prob) in zip(axes, mc_series):
    ax_prob = ax.twinx()
    positive_prob = np.where(abort_prob > 0, abort_prob, np.nan)

    infid_line, = ax.loglog(
        gamma_list,
        infid,
        '.-',
        color='C0',
        linewidth=2,
        label='Infidelity',
    )
    prob_line, = ax_prob.loglog(
        gamma_list,
        positive_prob,
        '.--',
        color='#A60628',
        linewidth=2,
        label=r'$p_\mathrm{abort}$',
    )

    ax.set_title(f'$m_c={mc}$', fontsize=13)
    ax.set_xlim([1e-3, 1e-1])
    ax.set_ylim([1, 1e-6])
    ax.grid(True, which='both', alpha=0.35)
    ax_prob.set_ylim([1, 1e-6])

    if ax in axes[::2]:
        ax.set_ylabel('Infidelity')
    if ax in axes[1::2]:
        ax_prob.set_ylabel(r'$p_\mathrm{abort}$')
    else:
        ax_prob.set_yticklabels([])
    if ax in axes[2:]:
        ax.set_xlabel(r'$\gamma$')

    ax.legend([infid_line, prob_line], ['Infidelity', r'$p_\mathrm{abort}$'], fontsize=10, loc='lower left')

fig.tight_layout()
fig.savefig(f'../plots/f_cut_appendix,{d=},{alpha=}.svg', bbox_inches='tight')
fig.savefig(f'../plots/f_cut_appendix,{d=},{alpha=}.pdf', bbox_inches='tight')
plt.show()
```
